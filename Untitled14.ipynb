```python
import json
import random
from IPython.display import display, HTML
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
import ipywidgets as widgets

# 1. Initialize the Qiskit Quantum Game Engine
class InteractiveQuantumPallanguzhi:
    def __init__(self):
        self.reset_game()
        self.sim = AerSimulator()

    def reset_game(self):
        self.board = [6] * 14
        self.scores = {"player": 0, "ai": 0}
        self.current_turn = "player"
        self.game_over = False

    def get_quantum_entropy(self):
        """Runs a single-qubit circuit to generate a 0 or 1 via measurement collapse."""
        qc = QuantumCircuit(1, 1)
        qc.h(0)
        qc.measure(0, 0)
        result = self.sim.run(qc, shots=1).result()
        counts = result.get_counts()
        return int(list(counts.keys())[0])

game_engine = InteractiveQuantumPallanguzhi()

# 2. Setup bidirectional messaging for standard Jupyter via HTML / JS / ipywidgets
# This invisible text widget acts as a bridge: JS writes to its value, Python reads and responds.
comm_widget = widgets.Text(value='', layout=widgets.Layout(display='none'))

def handle_msg_change(change):
    try:
        if not change['new']:
            return
        msg = json.loads(change['new'])
        if msg.get('action') == 'get_quantum_state':
            state = game_engine.get_quantum_entropy()
            # Send the state back to JavaScript by updating the value with a response structure
            comm_widget.value = json.dumps({"response_for": msg['id'], "quantum_state": state})
    except Exception as e:
        pass

comm_widget.observe(handle_msg_change, names='value')
display(comm_widget)

# 3. HTML Game Board with support for both environments
html_content = """
<div id="game-container" style="max-width: 800px; margin: auto; font-family: 'Segoe UI', Tahoma, Geneva, Verdana, sans-serif; background-color: #f7eedb; border: 12px solid #5c3a21; border-radius: 20px; padding: 25px; box-shadow: 0 10px 25px rgba(0,0,0,0.3); text-align: center;">
    <div style="display: flex; justify-content: center; align-items: center; margin-bottom: 20px;">
        <h2 style="color: #5c3a21; margin: 0; font-weight: bold;">பல்லாங்குழி - Quantum Pallanguzhi 🌌</h2>
    </div>

    <div id="status-panel" style="background-color: #eedebd; border-radius: 10px; padding: 12px; margin-bottom: 20px; border: 2px dashed #8b5a2b; font-size: 15px; min-height: 40px; color: #5c3a21;">
        Quantum Game Ready! Click a pit on your side (bottom row) to make a move.
    </div>

    <div style="background: radial-gradient(#a05a2c, #5c3a21); border-radius: 15px; padding: 25px; box-shadow: inset 0 0 20px #000, 0 0 15px rgba(0, 150, 255, 0.4); position: relative;">
        <!-- AI Side -->
        <div style="display: flex; justify-content: space-around; margin-bottom: 30px;">
            <div style="color: #ffd700; font-size: 12px; font-weight: bold; position: absolute; top: 5px; left: 25px;">AI Side (Quantum)</div>
            <div class="pit-container" id="pit-13"><div class="pit-label">Pit 7</div><div class="pit-cavity"><div class="stone-count">6</div></div></div>
            <div class="pit-container" id="pit-12"><div class="pit-label">Pit 6</div><div class="pit-cavity"><div class="stone-count">6</div></div></div>
            <div class="pit-container" id="pit-11"><div class="pit-label">Pit 5</div><div class="pit-cavity"><div class="stone-count">6</div></div></div>
            <div class="pit-container" id="pit-10"><div class="pit-label">Pit 4</div><div class="pit-cavity"><div class="stone-count">6</div></div></div>
            <div class="pit-container" id="pit-9"><div class="pit-label">Pit 3</div><div class="pit-cavity"><div class="stone-count">6</div></div></div>
            <div class="pit-container" id="pit-8"><div class="pit-label">Pit 2</div><div class="pit-cavity"><div class="stone-count">6</div></div></div>
            <div class="pit-container" id="pit-7"><div class="pit-label">Pit 1</div><div class="pit-cavity"><div class="stone-count">6</div></div></div>
        </div>

        <!-- Player Side -->
        <div style="display: flex; justify-content: space-around;">
            <div style="color: #ffd700; font-size: 12px; font-weight: bold; position: absolute; bottom: 5px; left: 25px;">Player Side (You)</div>
            <div class="pit-container active-pit" id="pit-0" onclick="makePlayerMove(0)"><div class="pit-label">Pit 1</div><div class="pit-cavity"><div class="stone-count">6</div></div></div>
            <div class="pit-container active-pit" id="pit-1" onclick="makePlayerMove(1)"><div class="pit-label">Pit 2</div><div class="pit-cavity"><div class="stone-count">6</div></div></div>
            <div class="pit-container active-pit" id="pit-2" onclick="makePlayerMove(2)"><div class="pit-label">Pit 3</div><div class="pit-cavity"><div class="stone-count">6</div></div></div>
            <div class="pit-container active-pit" id="pit-3" onclick="makePlayerMove(3)"><div class="pit-label">Pit 4</div><div class="pit-cavity"><div class="stone-count">6</div></div></div>
            <div class="pit-container active-pit" id="pit-4" onclick="makePlayerMove(4)"><div class="pit-label">Pit 5</div><div class="pit-cavity"><div class="stone-count">6</div></div></div>
            <div class="pit-container active-pit" id="pit-5" onclick="makePlayerMove(5)"><div class="pit-label">Pit 6</div><div class="pit-cavity"><div class="stone-count">6</div></div></div>
            <div class="pit-container active-pit" id="pit-6" onclick="makePlayerMove(6)"><div class="pit-label">Pit 7</div><div class="pit-cavity"><div class="stone-count">6</div></div></div>
        </div>
    </div>

    <!-- Score Board -->
    <div style="display: flex; justify-content: space-around; margin-top: 25px; background-color: #eedebd; border-radius: 10px; padding: 15px; border: 2px solid #5c3a21;">
        <div style="font-size: 18px; font-weight: bold; color: #5c3a21;">🏆 Player Score: <span id="player-score">0</span></div>
        <div style="font-size: 18px; font-weight: bold; color: #5c3a21;">🤖 AI Score: <span id="ai-score">0</span></div>
    </div>

    <!-- Control Dashboard -->
    <div style="margin-top: 20px; display: flex; justify-content: space-around;">
        <button onclick="restartGame()" style="background-color: #8b5a2b; color: white; border: none; padding: 10px 20px; font-size: 14px; border-radius: 5px; cursor: pointer; font-weight: bold;">🔄 New Game</button>
        <button onclick="triggerQuantumInterference()" style="background-color: #007acc; color: white; border: none; padding: 10px 20px; font-size: 14px; border-radius: 5px; cursor: pointer; font-weight: bold; box-shadow: 0 0 10px #007acc;">⚛️ Quantum Interference</button>
    </div>
</div>

<style>
    .pit-container { display: flex; flex-direction: column; align-items: center; width: 80px; transition: all 0.2s; }
    .pit-label { color: #ffd700; font-size: 14px; margin-bottom: 5px; font-weight: bold; }
    .pit-cavity { background: radial-gradient(#2c1505, #110500); width: 60px; height: 60px; border-radius: 50%; border: 4px solid #8b5a2b; display: flex; justify-content: center; align-items: center; }
    .active-pit { cursor: pointer; }
    .active-pit:hover .pit-cavity { border-color: #00bcff; box-shadow: 0 0 15px #00bcff; transform: scale(1.08); }
    .stone-count { color: #ffffff; font-size: 22px; font-weight: 900; text-shadow: -1px -1px 0 #000, 1px -1px 0 #000, -1px 1px 0 #000, 1px 1px 0 #000; }
</style>

<script>
let gameState = {
    board: [6, 6, 6, 6, 6, 6, 6, 6, 6, 6, 6, 6, 6, 6],
    scores: {player: 0, ai: 0},
    current_turn: 'player',
    gameOver: false
};

function updateUI(explanation = "") {
    document.getElementById('player-score').innerText = gameState.scores.player;
    document.getElementById('ai-score').innerText = gameState.scores.ai;
    for(let i=0; i<14; i++) {
        const pit = document.getElementById('pit-' + i);
        if(pit) { pit.querySelector('.stone-count').innerText = gameState.board[i]; }
    }
    const statusPanel = document.getElementById('status-panel');
    if (gameState.gameOver) {
        statusPanel.innerHTML = `<strong>Game Over! Final Scores -> You: ${gameState.scores.player} | AI: ${gameState.scores.ai}</strong>`;
    } else {
        statusPanel.innerHTML = explanation || (gameState.current_turn === 'player' ? "Your Turn! Choose an active pit or execute Quantum Interference!" : "🤖 AI is thinking...");
    }
}

function makePlayerMove(pitIndex) {
    if (gameState.current_turn !== 'player' || gameState.gameOver) return;
    if (gameState.board[pitIndex] === 0) return;
    executeMove(pitIndex);
}

function executeMove(startPit) {
    let board = [...gameState.board];
    let scores = {...gameState.scores};
    let player = startPit < 7 ? 'player' : 'ai';
    let stones = board[startPit];
    board[startPit] = 0;

    let currIdx = startPit;
    while(stones > 0) {
        currIdx = (currIdx + 1) % 14;
        board[currIdx]++;
        stones--;
    }

    let captured = 0;
    let nextIdx = (currIdx + 1) % 14;
    if (board[nextIdx] === 0) {
        let captureIdx = (nextIdx + 1) % 14;
        captured = board[captureIdx];
        board[captureIdx] = 0;
        scores[player] += captured;
    }

    gameState.board = board;
    gameState.scores = scores;
    gameState.current_turn = (player === 'player' ? 'ai' : 'player');

    checkGameOver();
    updateUI(`Distributed from Pit ${startPit + 1}. Captured ${captured}!`);

    if (gameState.current_turn === 'ai' && !gameState.gameOver) {
        setTimeout(triggerAIMove, 1000);
    }
}

function triggerAIMove() {
    let validMoves = [];
    for(let i=7; i<14; i++) { if (gameState.board[i] > 0) validMoves.push(i); }
    if (validMoves.length === 0) { gameState.gameOver = true; updateUI(); return; }
    let aiMove = validMoves[Math.floor(Math.random() * validMoves.length)];
    executeMove(aiMove);
}

function triggerQuantumInterference() {
    if (gameState.gameOver) return;
    document.getElementById('status-panel').innerText = "🌌 Executing Quantum Circuit...";

    const requestId = Math.random().toString(36).substring(2, 9);

    // Native Cross-Platform Jupyter / Colab communication:
    if (window.google && google.colab) {
        google.colab.kernel.invokeFunction('notebook.run_quantum_action', [], {}).then(function(result) {
            applyQuantumResult(result.data['application/json'].quantum_state);
        });
    } else {
        // Fallback to standard Jupyter Comm channel via hidden widget
        const widgetEl = document.querySelector('.widget-text input');
        if (widgetEl) {
            widgetEl.value = JSON.stringify({action: 'get_quantum_state', id: requestId});
            widgetEl.dispatchEvent(new Event('change'));

            const checkInterval = setInterval(() => {
                try {
                    const response = JSON.parse(widgetEl.value);
                    if (response.response_for === requestId) {
                        clearInterval(checkInterval);
                        applyQuantumResult(response.quantum_state);
                        widgetEl.value = '';
                    }
                } catch(e) {}
            }, 100);
        }
    }
}

function applyQuantumResult(collapseValue) {
    if (collapseValue === 1) {
        let validPits = [];
        for(let i=0; i<7; i++) { if (gameState.board[i] > 0) validPits.push(i); }
        if (validPits.length > 0) {
            let chosen = validPits[Math.floor(Math.random() * validPits.length)];
            gameState.board[chosen] += 2;
            updateUI(`⚛️ Quantum collapsed to |1>! Added 2 stones to Pit ${chosen + 1}.`);
        }
    } else {
        updateUI("⚛️ Quantum collapsed to |0>. No changes occurred.");
    }
}

function checkGameOver() {
    if (gameState.board.slice(0,7).every(x => x === 0) || gameState.board.slice(7,14).every(x => x === 0)) {
        gameState.gameOver = true;
    }
}

function restartGame() {
    gameState = { board: [6,6,6,6,6,6,6,6,6,6,6,6,6,6], scores: {player:0, ai:0}, current_turn: 'player', gameOver: false };
    updateUI("Quantum board reset!");
}

updateUI();
</script>
"""

display(HTML(html_content))
```
