```python
import json
import random
from IPython.display import HTML, display

# Game State & AI logic backend
class PallanguzhiGame:
    def __init__(self):
        self.reset_game()

    def reset_game(self):
        # Each player has 7 pits. Pits 0-6 are Player (Bottom), 7-13 are AI (Top, indexed counter-clockwise)
        # Traditional starting setup: 6 stones per pit
        self.board = [6] * 14
        self.scores = {"player": 0, "ai": 0}
        self.current_turn = "player"  # 'player' or 'ai'
        self.game_over = False

    def get_valid_moves(self, player):
        start, end = (0, 7) if player == "player" else (7, 14)
        return [i for i in range(start, end) if self.board[i] > 0]

    def simulate_move(self, board, scores, start_pit):
        # Returns (new_board, new_scores, next_turn, captured_stones, explanation)
        temp_board = list(board)
        temp_scores = dict(scores)
        current_player = "player" if start_pit < 7 else "ai"

        stones = temp_board[start_pit]
        temp_board[start_pit] = 0

        curr_idx = start_pit
        captured = 0

        # Classic distribution logic
        while stones > 0:
            curr_idx = (curr_idx + 1) % 14
            temp_board[curr_idx] += 1
            stones -= 1

        # Capture check: if last stone landed, check next pit
        next_idx = (curr_idx + 1) % 14
        if temp_board[next_idx] == 0:
            # Capture stones from the pit after the empty one
            capture_idx = (next_idx + 1) % 14
            captured = temp_board[capture_idx]
            temp_board[capture_idx] = 0
            temp_scores[current_player] += captured

        # Bonus turn if landing in own pit with even stones
        bonus_turn = False
        if current_player == "player" and curr_idx < 7 and temp_board[curr_idx] % 2 == 0:
            bonus_turn = True
        elif current_player == "ai" and curr_idx >= 7 and temp_board[curr_idx] % 2 == 0:
            bonus_turn = True

        next_turn = current_player if bonus_turn else ("ai" if current_player == "player" else "player")

        return temp_board, temp_scores, next_turn, captured

    def evaluate_board(self, board, scores):
        # Strategic evaluation: score difference + board control
        player_board = sum(board[0:7])
        ai_board = sum(board[7:14])
        return (scores["ai"] - scores["player"]) * 2.0 + (ai_board - player_board) * 0.5

    def minimax(self, board, scores, depth, alpha, beta, maximizing_player):
        if depth == 0:
            return self.evaluate_board(board, scores), None

        if maximizing_player:
            max_eval = float('-inf')
            best_move = None
            moves = [i for i in range(7, 14) if board[i] > 0]
            if not moves: return self.evaluate_board(board, scores), None
            for move in moves:
                nb, ns, nt, _ = self.simulate_move(board, scores, move)
                is_max = (nt == "ai")
                val, _ = self.minimax(nb, ns, depth - 1, alpha, beta, is_max)
                if val > max_eval:
                    max_eval = val
                    best_move = move
                alpha = max(alpha, val)
                if beta <= alpha:
                    break
            return max_eval, best_move
        else:
            min_eval = float('inf')
            best_move = None
            moves = [i for i in range(0, 7) if board[i] > 0]
            if not moves: return self.evaluate_board(board, scores), None
            for move in moves:
                nb, ns, nt, _ = self.simulate_move(board, scores, move)
                is_max = (nt == "ai")
                val, _ = self.minimax(nb, ns, depth - 1, alpha, beta, is_max)
                if val < min_eval:
                    min_eval = val
                    best_move = move
                beta = min(beta, val)
                if beta <= alpha:
                    break
            return min_eval, best_move

    def get_ai_move(self, difficulty="Medium"):
        depth = 1 if difficulty == "Easy" else (3 if difficulty == "Medium" else 5)
        _, move = self.minimax(self.board, self.scores, depth, float('-inf'), float('inf'), True)
        if move is None:
            valid = self.get_valid_moves("ai")
            if valid: move = random.choice(valid)
        return move

# Create game engine and render interactive game UI
game_engine = PallanguzhiGame()
```

```python
import json
import random
from IPython.display import HTML, display
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

# Advanced Pallanguzhi Engine utilizing Quantum Mechanics via Qiskit
class QuantumPallanguzhiGame:
    def __init__(self):
        self.reset_game()
        self.sim = AerSimulator()

    def reset_game(self):
        self.board = [6] * 14
        self.scores = {"player": 0, "ai": 0}
        self.current_turn = "player"
        self.game_over = False

    def get_quantum_entropy(self):
        """Uses a single-qubit Hadamard gate circuit to generate a true quantum random bit."""
        qc = QuantumCircuit(1, 1)
        qc.h(0)
        qc.measure(0, 0)
        result = self.sim.run(qc, shots=1).result()
        counts = result.get_counts()
        # Return 0 or 1 based on quantum collapse
        return int(list(counts.keys())[0])

    def simulate_move(self, board, scores, start_pit):
        temp_board = list(board)
        temp_scores = dict(scores)
        current_player = "player" if start_pit < 7 else "ai"

        stones = temp_board[start_pit]
        temp_board[start_pit] = 0

        curr_idx = start_pit
        captured = 0

        while stones > 0:
            curr_idx = (curr_idx + 1) % 14
            temp_board[curr_idx] += 1
            stones -= 1

        next_idx = (curr_idx + 1) % 14
        if temp_board[next_idx] == 0:
            capture_idx = (next_idx + 1) % 14
            captured = temp_board[capture_idx]
            temp_board[capture_idx] = 0
            temp_scores[current_player] += captured

        bonus_turn = False
        if current_player == "player" and curr_idx < 7 and temp_board[curr_idx] % 2 == 0:
            bonus_turn = True
        elif current_player == "ai" and curr_idx >= 7 and temp_board[curr_idx] % 2 == 0:
            bonus_turn = True

        next_turn = current_player if bonus_turn else ("ai" if current_player == "player" else "player")
        return temp_board, temp_scores, next_turn, captured

# Re-initialize the game engine
game_engine = QuantumPallanguzhiGame()
```

### 🌌 What is Quantum Interference?

Think of the **Quantum Interference** button as your "quantum wild card" during a match.

Normally, in traditional Pallanguzhi, every move is entirely classical and deterministic—you pick a pit, sow the seeds, and get a predictable outcome.

When you click **Quantum Interference**, the game executes a real quantum simulation using **Qiskit**:

1. **The Setup**: It initializes 1 qubit (a quantum bit).
2. **The Superposition**: It passes this qubit through a **Hadamard Gate** (`H`). This suspends the qubit in a physical state of limbo—it is simultaneously both $|0\rangle$ and $|1\rangle$ (a superposition state).
3. **The Measurement**: The simulation measures the qubit, forcing it to instantly collapse into either classical `0` or `1` with a 50/50 probability.

#### 🎁 The In-Game Rewards:
* **If it collapses to $|1\rangle$ (Success - 50% chance)**: The quantum state creates positive interference! It instantly injects **2 extra bonus stones** into one of your active pits at random. This can completely change your strategy, giving you extra ammunition to sweep your opponent's side.
* **If it collapses to $|0\rangle$ (Decoherence - 50% chance)**: The quantum state decoheres (gets lost to the environment). No bonus is given, and the board remains exactly as it was.
