```python
import json
import uuid

from IPython.display import HTML, JSON, display
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator


class QuantumEngine:
    """Builds the circuits and returns batches of measured results."""

    MAX_BATCH = 1024

    def __init__(self):
        self.sim = AerSimulator()

    @staticmethod
    def pit_circuit():
        """3 qubits in uniform superposition -> integer 0..7."""
        qc = QuantumCircuit(3, 3, name="pit_selector")
        qc.h([0, 1, 2])
        qc.measure([0, 1, 2], [0, 1, 2])
        return qc

    @staticmethod
    def link_circuit():
        """Bell pair: qubit 0 = your pit, qubit 1 = the opposite pit."""
        qc = QuantumCircuit(2, 2, name="quantum_link")
        qc.h(0)
        qc.cx(0, 1)
        qc.measure([0, 1], [0, 1])
        return qc

    def sample(self, kind, n):
        """One measurement per shot. 'pit' -> ints 0..7, 'link' -> [bit_pit, bit_opposite]."""
        n = max(1, min(int(n), self.MAX_BATCH))
        if kind == "pit":
            qc = self.pit_circuit()
        elif kind == "link":
            qc = self.link_circuit()
        else:
            raise ValueError(f"unknown sample kind: {kind!r}")
        memory = self.sim.run(transpile(qc, self.sim), shots=n, memory=True).result().get_memory()
        if kind == "pit":
            return [int(s, 2) for s in memory]
        # Qiskit bitstrings are little-endian: s[-1] is qubit 0.
        return [[int(s[-1]), int(s[-2])] for s in memory]


engine = QuantumEngine()


def _service(kind, n):
    return {"kind": kind, "samples": engine.sample(kind, n)}


def _register_colab():
    try:
        from google.colab import output
    except ImportError:
        return False

    def _callback(kind, n):
        return JSON(_service(kind, n))  # reaches the page as result.data["application/json"]

    output.register_callback("quantum_pallanguzhi.sample", _callback)
    return True


IN_COLAB = _register_colab()

PAGE_TEMPLATE = r"""<link rel="preconnect" href="https://fonts.googleapis.com">
<link rel="preconnect" href="https://fonts.gstatic.com" crossorigin>
<link rel="stylesheet" href="https://fonts.googleapis.com/css2?family=Fraunces:opsz,wght@9..144,600&family=Hanken+Grotesk:wght@400;500;700&family=JetBrains+Mono:wght@400;600&family=Noto+Serif+Tamil:wght@600&display=swap">
<style>
/* Layout: wooden board on the left, quantum lab on the right; stacks to one column on phones.
   Deliberately a single dark look (teak and brass), with teal reserved for anything quantum. */
.qp-app {
  --ground: #1b110a;
  --teak: #4a2c18;
  --teak-hi: #613b21;
  --well: #130a05;
  --well-rim: #8a5f2c;
  --brass: #d3a54d;
  --seed: #8c3d1b;
  --ink: #f2e6cf;
  --muted: #b8a58a;
  --panel: #251810;
  --line: #3d2a1b;
  --quantum: #5fd6c8;
  --quantum-ink: #06231f;
  --warn: #e98a5b;
  --font-display: "Fraunces", Georgia, "Times New Roman", serif;
  --font-body: "Hanken Grotesk", system-ui, -apple-system, "Segoe UI", sans-serif;
  --font-mono: "JetBrains Mono", ui-monospace, "SF Mono", Menlo, Consolas, monospace;
  --font-tamil: "Noto Serif Tamil", "Latha", serif;
  color-scheme: dark;
}
.qp-app, .qp-app * { box-sizing: border-box; }
.qp-app {
  background: var(--ground);
  color: var(--ink);
  font-family: var(--font-body);
  font-size: 15px;
  line-height: 1.5;
  padding: clamp(14px, 3vw, 28px);
  border-radius: 14px;
}
.wrap { max-width: 1080px; margin-inline: auto; display: flex; flex-direction: column; gap: 18px; }
header { display: flex; flex-wrap: wrap; align-items: baseline; gap: 4px 16px; }
h1 { margin: 0; font-family: var(--font-tamil); font-size: clamp(28px, 6vw, 40px); font-weight: 600; letter-spacing: 0.01em; line-height: 1.2; text-wrap: balance; }
.sub { font-family: var(--font-display); font-size: clamp(17px, 3vw, 22px); color: var(--brass); }
.status {
  display: flex; align-items: center; gap: 12px; min-height: 52px;
  background: var(--panel); border: 1px solid var(--line); border-radius: 10px; padding: 10px 14px;
}
.turn {
  flex: none; font-family: var(--font-mono); font-size: 11px; letter-spacing: 0.08em; text-transform: uppercase;
  padding: 4px 8px; border-radius: 4px; background: var(--brass); color: #2a1a0a; font-weight: 600;
}
.turn.ai { background: var(--quantum); color: var(--quantum-ink); }
.turn.done { background: var(--muted); color: #2a1a0a; }
.status p { margin: 0; min-width: 0; }

.layout { display: grid; grid-template-columns: minmax(0, 1fr); gap: 18px; align-items: start; }
@media (min-width: 880px) { .layout { grid-template-columns: minmax(0, 1.55fr) minmax(0, 1fr); } }
.col { display: flex; flex-direction: column; gap: 14px; min-width: 0; }

.board {
  background: linear-gradient(160deg, var(--teak-hi), var(--teak));
  border: 2px solid #2b1a0d; border-radius: 16px;
  padding: clamp(10px, 2.4vw, 22px);
  box-shadow: inset 0 2px 0 rgba(255,255,255,0.05), 0 10px 24px rgba(0,0,0,0.45);
  display: flex; flex-direction: column; gap: 8px;
}
.side-label { font-family: var(--font-mono); font-size: 11px; letter-spacing: 0.08em; text-transform: uppercase; color: #e4cf9f; opacity: 0.85; }
.side-label.right { text-align: right; }
.row { display: grid; grid-template-columns: repeat(7, minmax(0, 1fr)); gap: clamp(5px, 1.6vw, 14px); }
.row + .row { margin-top: clamp(4px, 1.5vw, 12px); }

.pit {
  position: relative; display: flex; flex-direction: column; align-items: center; gap: 4px;
  background: none; border: 0; padding: 0; margin: 0; color: inherit; font: inherit; min-width: 0;
}
.well {
  position: relative; display: block; width: 100%; aspect-ratio: 1; border-radius: 50%;
  background: radial-gradient(circle at 50% 38%, #24130a, var(--well) 70%);
  border: clamp(2px, 0.5vw, 4px) solid var(--well-rim);
  box-shadow: inset 0 4px 10px rgba(0,0,0,0.8);
  transition: box-shadow 0.18s, border-color 0.18s, transform 0.18s;
}
.seeds { position: absolute; inset: 6%; display: block; }
.seed {
  position: absolute; width: 17%; aspect-ratio: 1; border-radius: 50%;
  box-shadow: inset -1px -2px 2px rgba(0,0,0,0.5), 0 1px 1px rgba(0,0,0,0.6);
}
.badge {
  position: absolute; top: -4px; right: -2px; z-index: 2; min-width: 20px; padding: 1px 5px; text-align: center;
  font-family: var(--font-mono); font-weight: 600; font-size: clamp(10px, 2.4vw, 13px);
  background: var(--ground); color: var(--ink); border: 1px solid var(--well-rim); border-radius: 999px;
  font-variant-numeric: tabular-nums;
}
.plabel { font-family: var(--font-mono); font-size: 11px; color: #e4cf9f; opacity: 0.8; }
.pit.mine:not(:disabled) { cursor: pointer; }
.pit.mine:not(:disabled) .well { border-color: var(--brass); }
.pit.mine:not(:disabled):hover .well, .pit.mine:not(:disabled):focus-visible .well { box-shadow: inset 0 4px 10px rgba(0,0,0,0.8), 0 0 0 3px rgba(211,165,77,0.55); transform: scale(1.05); }
.pit.mine:focus-visible { outline: none; }
.pit.mine:disabled .well { opacity: 0.92; }
.pit.drop .well { box-shadow: inset 0 4px 10px rgba(0,0,0,0.8), 0 0 0 3px rgba(242,230,207,0.6); transform: scale(1.06); }
.pit.take .well { box-shadow: inset 0 4px 10px rgba(0,0,0,0.8), 0 0 0 3px rgba(233,138,91,0.8); }
.pit.grab .well { box-shadow: inset 0 4px 10px rgba(0,0,0,0.8), 0 0 0 4px var(--warn); transform: scale(1.06); }
.pit.qglow .well { border-color: var(--quantum); box-shadow: inset 0 4px 10px rgba(0,0,0,0.8), 0 0 0 4px rgba(95,214,200,0.55), 0 0 22px var(--quantum); transform: scale(1.08); }

.scores { display: grid; grid-template-columns: 1fr 1fr; gap: 10px; }
.score { background: var(--panel); border: 1px solid var(--line); border-radius: 10px; padding: 10px 14px; display: flex; justify-content: space-between; align-items: baseline; gap: 8px; }
.score span:first-child { color: var(--muted); font-size: 13px; }
.score b { font-family: var(--font-display); font-size: 30px; font-variant-numeric: tabular-nums; line-height: 1; }

.controls { display: flex; flex-wrap: wrap; gap: 10px; }
.btn {
  font: inherit; font-weight: 700; font-size: 14px; border-radius: 8px; padding: 11px 16px; cursor: pointer;
  border: 1px solid var(--brass); background: transparent; color: var(--brass);
}
.btn:hover:not(:disabled) { background: rgba(211,165,77,0.14); }
.btn.quantum { background: var(--quantum); border-color: var(--quantum); color: var(--quantum-ink); flex: 1 1 220px; }
.btn.quantum:hover:not(:disabled) { background: #7fe3d7; }
.btn:disabled { opacity: 0.4; cursor: not-allowed; }
.btn:focus-visible, summary:focus-visible { outline: 3px solid var(--ink); outline-offset: 2px; }
.hint { margin: 0; color: var(--muted); font-size: 13px; }

.lab { background: var(--panel); border: 1px solid var(--line); border-radius: 14px; padding: 16px; display: flex; flex-direction: column; gap: 18px; min-width: 0; }
.lab h2 { margin: 0; font-family: var(--font-display); font-size: 20px; font-weight: 600; }
.lab h3 { margin: 0 0 6px; font-size: 12px; letter-spacing: 0.08em; text-transform: uppercase; color: var(--quantum); font-family: var(--font-mono); font-weight: 600; }
.lab p { margin: 0 0 8px; color: var(--muted); font-size: 13px; }
pre.circuit { margin: 0 0 10px; padding: 10px 12px; overflow-x: auto; background: var(--well); border: 1px solid var(--line); border-radius: 8px; font-family: var(--font-mono); font-size: 13px; line-height: 1.35; color: var(--ink); }
.hist { display: grid; grid-template-columns: repeat(8, minmax(0, 1fr)); gap: 4px; align-items: end; height: 84px; }
.hcol { display: flex; flex-direction: column; justify-content: flex-end; align-items: center; height: 100%; gap: 3px; min-width: 0; }
.hbar { width: 100%; min-height: 2px; background: var(--quantum); border-radius: 2px 2px 0 0; transition: height 0.25s; }
.hcol.void .hbar { background: repeating-linear-gradient(45deg, var(--muted) 0 3px, transparent 3px 6px); }
.hnum { font-family: var(--font-mono); font-size: 10px; color: var(--muted); font-variant-numeric: tabular-nums; }
.hlab { display: grid; grid-template-columns: repeat(8, minmax(0, 1fr)); gap: 4px; margin-top: 4px; }
.hlab span { font-family: var(--font-mono); font-size: 9px; color: var(--muted); text-align: center; }
.tally { font-family: var(--font-mono); font-size: 12px; color: var(--ink); font-variant-numeric: tabular-nums; }
.log { list-style: none; margin: 0; padding: 0; display: flex; flex-direction: column; gap: 6px; }
.log li { display: grid; grid-template-columns: auto auto minmax(0, 1fr); gap: 8px; align-items: baseline; font-size: 13px; }
.tag { font-family: var(--font-mono); font-size: 10px; letter-spacing: 0.06em; padding: 2px 6px; border-radius: 4px; background: var(--line); color: var(--ink); }
.tag.link { background: var(--quantum); color: var(--quantum-ink); }
.ket { font-family: var(--font-mono); color: var(--quantum); }
.empty { color: var(--muted); font-size: 13px; }
details { background: var(--panel); border: 1px solid var(--line); border-radius: 10px; padding: 12px 14px; color: var(--muted); font-size: 13px; }
summary { cursor: pointer; color: var(--ink); font-weight: 700; }
details p { margin: 10px 0 0; }
@media (prefers-reduced-motion: reduce) { * { transition: none !important; } }
</style>
<div class="qp-app" id="qp-root-__UID__">
<div class="wrap">
  <header>
    <h1>பல்லாங்குழி</h1>
    <span class="sub">Quantum Pallanguzhi</span>
  </header>

  <div class="status" role="status" aria-live="polite">
    <span class="turn" id="turn">Your turn</span>
    <p id="statusText"></p>
  </div>

  <div class="layout">
    <section class="col" aria-label="Game board">
      <div class="board">
        <div class="side-label">Opponent · picks pits with 3 measured qubits</div>
        <div class="row" id="rowAI"></div>
        <div class="row" id="rowYou"></div>
        <div class="side-label right">You · tap a lit pit to sow</div>
      </div>

      <div class="scores">
        <div class="score"><span>You</span><b id="scoreYou">0</b></div>
        <div class="score"><span>Opponent</span><b id="scoreAI">0</b></div>
      </div>

      <div class="controls">
        <button class="btn quantum" id="linkBtn" type="button">Quantum Link (once per turn)</button>
        <button class="btn" id="newBtn" type="button">New game</button>
      </div>
      <p class="hint">Sow stones counter-clockwise. If the pit right after your last stone is empty, you capture the pit after it. Quantum Link picks one of your pits and entangles it with the opposite pit: both gain 2 stones, or neither does.</p>
    </section>

    <aside class="lab" aria-label="Quantum lab">
      <h2>Quantum lab</h2>
      <div class="tally" id="source"></div>

      <section>
        <h3>Pit register · 3 qubits</h3>
        <pre class="circuit" aria-label="Circuit: Hadamard on three qubits, then measure">q0: ─H─M
q1: ─H─M
q2: ─H─M</pre>
        <p>Each measurement gives 0–7 with equal odds. Seven is not a pit, so it is thrown back and measured again.</p>
        <div class="hist" id="hist"></div>
        <div class="hlab" id="histLab"></div>
      </section>

      <section>
        <h3>Quantum Link · Bell pair</h3>
        <pre class="circuit" aria-label="Circuit: Hadamard on qubit 0, controlled-NOT to qubit 1, then measure">q0: ─H──●──M   your pit
q1: ────X──M   opposite pit</pre>
        <div class="tally" id="tally">No pairs measured yet.</div>
      </section>

      <section>
        <h3>Measurement log</h3>
        <ul class="log" id="log"><li class="empty">Measurements show up here as the game goes.</li></ul>
      </section>
    </aside>
  </div>

  <details>
    <summary>Where the quantum results come from</summary>
    <p>Each result is one measurement shot from Qiskit's AerSimulator, running in this Colab runtime. The page asks Python for fresh batches while you play. AerSimulator is a classical simulator, so this is not a run on quantum hardware. If you restart the runtime, run the game cell again.</p>
  </details>
</div>
</div>
<script>
(function () {
  "use strict";

  // ---------- Qiskit results arrive from Python ----------
  var POOL = __POOL__;
  var buf = { pit: POOL.pit.slice(), link: POOL.link.slice() };
  var refilling = { pit: false, link: false };
  var liveBatches = 0, fallbackUsed = false;
  var transport = (window.google && google.colab && google.colab.kernel) ? "colab"
    : (window.Jupyter && Jupyter.notebook && Jupyter.notebook.kernel && Jupyter.notebook.kernel.comm_manager) ? "comm"
    : "none";

  function requestSamples(kind, n) {
    if (transport === "colab") {
      return google.colab.kernel.invokeFunction("quantum_pallanguzhi.sample", [kind, n], {})
        .then(function (r) { return r.data["application/json"].samples; });
    }
    if (transport === "comm") {
      return new Promise(function (resolve, reject) {
        var comm = Jupyter.notebook.kernel.comm_manager.new_comm("quantum_pallanguzhi", { kind: kind, n: n });
        var timer = setTimeout(function () { try { comm.close(); } catch (e) {} reject(new Error("timeout")); }, 6000);
        comm.on_msg(function (msg) {
          clearTimeout(timer);
          try { comm.close(); } catch (e) {}
          resolve((msg.content.data && msg.content.data.samples) || []);
        });
      });
    }
    return Promise.reject(new Error("no live bridge"));
  }
  function refill(kind) {
    if (transport === "none" || refilling[kind]) return;
    refilling[kind] = true;
    requestSamples(kind, kind === "pit" ? 256 : 64)
      .then(function (s) { Array.prototype.push.apply(buf[kind], s); liveBatches++; renderLab(); })
      .catch(function () {})
      .then(function () { refilling[kind] = false; });
  }
  // Never blocks the game: reads a stored result, tops the supply up in the background.
  function draw(kind) {
    if (buf[kind].length < 48) refill(kind);
    if (buf[kind].length) return buf[kind].shift();
    fallbackUsed = true;
    if (kind === "pit") return Math.floor(Math.random() * 8);
    var bit = Math.random() < 0.5 ? 1 : 0;
    return [bit, bit];
  }
  function sourceText() {
    var t = "Source: Qiskit AerSimulator, ";
    if (transport === "colab") t += "live from this Colab runtime (" + liveBatches + " refill" + (liveBatches === 1 ? "" : "s") + ")";
    else if (transport === "comm") t += "live from this Jupyter kernel (" + liveBatches + " refills)";
    else t += "measured when the cell ran";
    t += " · results in stock: " + buf.pit.length + " pit, " + buf.link.length + " link";
    if (fallbackUsed) t += " · stock ran out, some results came from Math.random";
    return t;
  }

  var pitCounts = [0, 0, 0, 0, 0, 0, 0, 0];
  var linkStats = { total: 0, matched: 0, both1: 0 };
  var logItems = [];

  function measurePit() {
    var v = draw("pit");
    pitCounts[v]++;
    return v;
  }
  function measureLink() {
    var pair = draw("link"), a = pair[0], b = pair[1];
    linkStats.total++;
    if (a === b) linkStats.matched++;
    if (a === 1 && b === 1) linkStats.both1++;
    return [a, b];
  }

  // ---------- DOM ----------
  var ROOT = document.getElementById("qp-root-__UID__");
  var $ = function (id) { return ROOT.querySelector("#" + id); };
  var REDUCED = window.matchMedia && matchMedia("(prefers-reduced-motion: reduce)").matches;
  function sleep(ms) { return new Promise(function (r) { setTimeout(r, REDUCED ? Math.min(ms, 20) : ms); }); }
  var pits = [];

  function makePit(i, mine) {
    var el = document.createElement(mine ? "button" : "div");
    el.className = "pit " + (mine ? "mine" : "theirs");
    if (mine) {
      el.type = "button";
      el.setAttribute("aria-label", "Your pit " + (i + 1));
      el.addEventListener("click", function () { onPit(i); });
    }
    el.innerHTML = '<span class="well"><span class="seeds"></span></span><span class="badge">6</span><span class="plabel">' + (mine ? i + 1 : i - 6) + "</span>";
    pits[i] = el;
    return el;
  }
  (function buildBoard() {
    var i;
    for (i = 13; i >= 7; i--) $("rowAI").appendChild(makePit(i, false));
    for (i = 0; i < 7; i++) $("rowYou").appendChild(makePit(i, true));
    var hist = $("hist"), lab = $("histLab");
    for (i = 0; i < 8; i++) {
      var col = document.createElement("div");
      col.className = "hcol" + (i === 7 ? " void" : "");
      col.title = i === 7 ? "111 = 7 is not a pit, so it is re-measured" : "000-110 map to pits 1-7";
      col.innerHTML = '<span class="hnum">0</span><div class="hbar" style="height:2px"></div>';
      hist.appendChild(col);
      var l = document.createElement("span");
      l.textContent = i.toString(2).padStart(3, "0");
      lab.appendChild(l);
    }
  })();

  function renderSeeds(el, count) {
    var box = el.querySelector(".seeds"), n = Math.min(count, 30), html = "", k;
    var rmax = Math.min(1, 0.3 + n * 0.07);
    for (k = 0; k < n; k++) {
      var r = n > 1 ? Math.sqrt((k + 0.5) / n) * rmax : 0;
      var ang = k * 2.399963;
      var x = 50 + r * Math.cos(ang) * 40 - 8.5, y = 50 + r * Math.sin(ang) * 40 - 8.5;
      var hue = 12 + ((k * 7) % 16), light = 24 + ((k * 5) % 10);
      html += '<i class="seed" style="left:' + x.toFixed(1) + "%;top:" + y.toFixed(1) + "%;background:radial-gradient(circle at 35% 30%, hsl(" + hue + ",60%," + (light + 18) + "%), hsl(" + hue + ",62%," + light + '%))"></i>';
    }
    box.innerHTML = html;
  }

  // ---------- game state ----------
  var S;
  function fresh() {
    return { board: [6, 6, 6, 6, 6, 6, 6, 6, 6, 6, 6, 6, 6, 6], scores: { you: 0, ai: 0 }, turn: "you", over: false, busy: false, linkUsed: false };
  }
  function setStatus(text) { $("statusText").textContent = text; }

  function renderPit(i) {
    var el = pits[i], c = S.board[i];
    if (el._c !== c) { renderSeeds(el, c); el._c = c; el.querySelector(".badge").textContent = c; }
  }
  function render() {
    var i;
    for (i = 0; i < 14; i++) renderPit(i);
    $("scoreYou").textContent = S.scores.you;
    $("scoreAI").textContent = S.scores.ai;
    var canPlay = S.turn === "you" && !S.over && !S.busy;
    for (i = 0; i < 7; i++) pits[i].disabled = !(canPlay && S.board[i] > 0);
    $("linkBtn").disabled = !(canPlay && !S.linkUsed);
    var t = $("turn");
    t.className = "turn" + (S.over ? " done" : S.turn === "ai" ? " ai" : "");
    t.textContent = S.over ? "Game over" : S.turn === "you" ? "Your turn" : "Opponent";
  }
  function flash(idx, cls, ms) {
    idx.forEach(function (i) { pits[i].classList.add(cls); });
    setTimeout(function () { idx.forEach(function (i) { pits[i].classList.remove(cls); }); }, ms);
  }

  // ---------- lab rendering ----------
  function renderLab() {
    var max = Math.max.apply(null, pitCounts.concat([1])), cols = $("hist").children, i;
    for (i = 0; i < 8; i++) {
      cols[i].querySelector(".hnum").textContent = pitCounts[i];
      cols[i].querySelector(".hbar").style.height = Math.max(2, Math.round((pitCounts[i] / max) * 60)) + "px";
    }
    $("tally").textContent = linkStats.total
      ? linkStats.total + " pair" + (linkStats.total > 1 ? "s" : "") + " measured · " + linkStats.matched + " matched · " + (linkStats.total - linkStats.matched) + " mismatched · |11⟩ " + linkStats.both1 + " times"
      : "No pairs measured yet.";
    $("source").textContent = sourceText();
    var log = $("log");
    if (!logItems.length) return;
    log.innerHTML = "";
    logItems.slice(0, 6).forEach(function (e) {
      var li = document.createElement("li");
      var tag = document.createElement("span"); tag.className = "tag" + (e.kind === "LINK" ? " link" : ""); tag.textContent = e.kind;
      var ket = document.createElement("span"); ket.className = "ket"; ket.textContent = "|" + e.bits + "⟩";
      var txt = document.createElement("span"); txt.textContent = e.text;
      li.appendChild(tag); li.appendChild(ket); li.appendChild(txt);
      log.appendChild(li);
    });
  }
  function addLog(kind, bits, text) {
    logItems.unshift({ kind: kind, bits: bits, text: text });
    renderLab();
  }

  // ---------- rules ----------
  function sum(a, b) { var t = 0, i; for (i = a; i < b; i++) t += S.board[i]; return t; }
  function checkOver() {
    if (sum(0, 7) === 0 || sum(7, 14) === 0) {
      S.scores.you += sum(0, 7);
      S.scores.ai += sum(7, 14);
      S.board = S.board.map(function () { return 0; });
      S.over = true;
      return true;
    }
    return false;
  }
  function endMessage() {
    var y = S.scores.you, a = S.scores.ai;
    return (y > a ? "You win " : y < a ? "The opponent wins " : "Draw ") + y + " to " + a + ". Leftover stones went to the side they sat on.";
  }

  function onPit(i) {
    if (S.turn !== "you" || S.over || S.busy || S.board[i] === 0) return;
    playMove(i);
  }

  async function playMove(start) {
    var st = S, who = start < 7 ? "you" : "ai", label = who === "you" ? start + 1 : start - 6;
    st.busy = true; render();
    var stones = st.board[start];
    st.board[start] = 0; renderPit(start);
    flash([start], "grab", 380);
    setStatus((who === "you" ? "You lift " : "Opponent lifts ") + stones + " stones from pit " + label + ".");
    await sleep(420); if (st !== S) return;

    var idx = start, k;
    for (k = 0; k < stones; k++) {
      idx = (idx + 1) % 14;
      st.board[idx]++; renderPit(idx);
      flash([idx], "drop", 150);
      await sleep(150); if (st !== S) return;
    }

    var next = (idx + 1) % 14, captured = 0, msg;
    if (st.board[next] === 0) {
      var cap = (next + 1) % 14;
      captured = st.board[cap];
      if (captured > 0) {
        flash([cap], "take", 700);
        await sleep(450); if (st !== S) return;
        st.board[cap] = 0; st.scores[who] += captured; renderPit(cap);
      }
    }
    msg = (who === "you" ? "You sowed" : "Opponent sowed") + " pit " + label + (captured ? " and captured " + captured + "." : ".");

    st.busy = false;
    st.turn = who === "you" ? "ai" : "you";
    if (st.turn === "you") st.linkUsed = false;

    if (checkOver()) { render(); setStatus(msg + " " + endMessage()); return; }
    render();
    if (st.turn === "ai") { setStatus(msg + " The opponent is measuring its qubits."); aiTurn(st); }
    else setStatus(msg + " Your turn.");
  }

  async function aiTurn(st) {
    st.busy = true; render();
    await sleep(900); if (st !== S) return;
    var pick = null, tries = 0, bits = "", i;
    while (pick === null && tries < 60) {
      tries++;
      var v = measurePit();
      if (v < 7 && st.board[7 + v] > 0) { pick = 7 + v; bits = v.toString(2).padStart(3, "0"); }
    }
    if (pick === null) {
      for (i = 7; i < 14; i++) if (st.board[i] > 0) { pick = i; break; }
      bits = "---";
      addLog("AI", bits, "No legal pit after 60 measurements, took pit " + (pick - 6) + ".");
    } else {
      addLog("AI", bits, "pit " + (pick - 6) + (tries > 1 ? " after " + (tries - 1) + " re-measure" + (tries > 2 ? "s" : "") : ""));
    }
    st.busy = false;
    playMove(pick);
  }

  function quantumLink() {
    if (S.turn !== "you" || S.over || S.busy || S.linkUsed) return;
    var p = null, tries = 0, i;
    while (p === null && tries < 60) {
      tries++;
      var v = measurePit();
      if (v < 7 && S.board[v] > 0) p = v;
    }
    if (p === null) { for (i = 0; i < 7; i++) if (S.board[i] > 0) { p = i; break; } }
    if (p === null) return;
    var m = 13 - p, pair = measureLink(), a = pair[0], b = pair[1];
    S.linkUsed = true;
    S.board[p] += 2 * a; S.board[m] += 2 * b;
    render();
    flash([p, m], "qglow", 1800);
    var bits = "" + b + a, text, status;
    if (a === 1 && b === 1) {
      text = "your pit " + (p + 1) + " and opposite pit " + (m - 6) + " both gain 2";
      status = "Entangled pair collapsed to |11⟩: your pit " + (p + 1) + " and the opposite pit " + (m - 6) + " each gained 2 stones. Now pick a pit.";
    } else if (a === 0 && b === 0) {
      text = "pits " + (p + 1) + " and " + (m - 6) + " both unchanged";
      status = "Entangled pair collapsed to |00⟩: pit " + (p + 1) + " and the opposite pit stay as they were. Now pick a pit.";
    } else {
      text = "outcomes diverged";
      status = "The pair gave different results, which a Bell state should not do.";
    }
    addLog("LINK", bits, text);
    setStatus(status);
  }

  function newGame() {
    S = fresh();
    pits.forEach(function (el) { el._c = -1; });
    render();
    setStatus("Six stones in every pit. Tap one of your pits to start, or use Quantum Link first.");
  }

  $("linkBtn").addEventListener("click", quantumLink);
  $("newBtn").addEventListener("click", newGame);
  renderLab();
  newGame();
})();
</script>
"""


def build_page():
    uid = uuid.uuid4().hex[:8]
    pool = {"pit": engine.sample("pit", 512), "link": engine.sample("link", 128)}
    return PAGE_TEMPLATE.replace("__UID__", uid).replace("__POOL__", json.dumps(pool))


display(HTML(build_page()))
if not IN_COLAB:
    print("Not running in Colab: the game works, but uses the results measured when this cell ran.")
```
